### Read the current state of the element analysis bcalm table: 
- Expecting: Table of the element based BCalm results using a scrambled control (e.g. "bcalm_adj_p_NGN2") and a column "is_considered_open_acc_SCREEN" which is annotated based on data from e.g. SCREEN information for H1, H9, NPC, neural stem cell and excitatory neuron

```python
import pandas as pd
```

```python
element_metadata_path = "/home/kisa/coding/80K_MPRA/element_variant_analysis_output/ngn2_wtc11_elements_bcalm_scrambled095_screenv4_annotation_groups_2507_ngn2_neuro_ctrl.tsv.gz"
element_metadata_table = pd.read_csv(element_metadata_path, sep="\t")
element_metadata_table.columns
```

```python
# annotate the expected column:
columns_of_interest_for_sequences_and_screen = ['name', 'chr', 'start', 'end', 'strand', 'sequence', 'raw_activity_NGN2_log2FC_bcalm', 'bcalm_adj_p_NGN2', 'normalized_activity_NGN2', 'SCREEN_ID', 'is_NPC_SCREEN_cCRE', 'is_H9_SCREEN_cCRE', 'is_H1_SCREEN_cCRE', 'is_neural_stem_SCREEN_cCRE', 'is_excitatory_neuron_SCREEN_cCRE']
element_metadata_table = element_metadata_table[columns_of_interest_for_sequences_and_screen].copy()

element_metadata_table['is_considered_open_acc_SCREEN'] = element_metadata_table[['is_NPC_SCREEN_cCRE', 'is_neural_stem_SCREEN_cCRE', 'is_excitatory_neuron_SCREEN_cCRE']].any(axis=1)
element_metadata_table
```

```python
ref_only_element_metadata_table = element_metadata_table.loc[element_metadata_table['name'].str.contains("cardiac_neuro_cava_random:") & ~element_metadata_table['name'].str.contains(":ALT_")].copy()
ref_only_element_metadata_table['start'] = ref_only_element_metadata_table['start'].astype(int)
ref_only_element_metadata_table['end'] = ref_only_element_metadata_table['end'].astype(int)
ref_only_element_metadata_table
```

```python
# add column indicating quantitative readout:
ref_only_element_metadata_table['has_measurements'] = ref_only_element_metadata_table['raw_activity_NGN2_log2FC_bcalm'].notna()
ref_only_element_metadata_table['has_measurements'].value_counts()

# add column indicating significant elements:
ref_only_element_metadata_table['is_significant'] = (ref_only_element_metadata_table['bcalm_adj_p_NGN2'] < 0.05) & ref_only_element_metadata_table['has_measurements']
ref_only_element_metadata_table['is_significant'].value_counts()
```

```python
ref_only_element_metadata_table_with_measurements = ref_only_element_metadata_table.loc[ref_only_element_metadata_table['has_measurements']].copy()
ref_only_element_metadata_table_with_measurements
print(f"Total elements with measurements: {ref_only_element_metadata_table_with_measurements.shape[0]}")
```

### Overview of the elements and their activity in numbers: 
- 25578 total elements
- How many are significantly active? 3091
- How many are significantly more active? 3091
- How many are significantly less active? 0
- How many have no significant activity? 22487

```python
ref_only_element_metadata_table_with_measurements['activity_category'] = ref_only_element_metadata_table_with_measurements.apply(lambda row: 'more_active' if (row['is_significant'] and row['raw_activity_NGN2_log2FC_bcalm'] > 0) else ('less_active' if (row['is_significant'] and row['raw_activity_NGN2_log2FC_bcalm'] < 0) else 'not_significant'), axis=1)
ref_only_element_metadata_table_with_measurements['activity_category'].value_counts()
```

```python
ref_only_element_metadata_table_with_measurements_duplicated = ref_only_element_metadata_table_with_measurements.loc[ref_only_element_metadata_table_with_measurements.duplicated(subset=['chr', 'start', 'end'], keep=False)].sort_values(by=['chr', 'start', 'end'])
ref_only_element_metadata_table_with_measurements_deduplicated = ref_only_element_metadata_table_with_measurements_duplicated.loc[ref_only_element_metadata_table_with_measurements_duplicated['name'].str.contains("_fwd_")].copy()
if ref_only_element_metadata_table_with_measurements_deduplicated.loc[ref_only_element_metadata_table_with_measurements_deduplicated.duplicated(subset=['chr', 'start', 'end'], keep=False)].shape[0] != 0:
    raise ValueError("There are still duplicated elements after filtering for _fwd_ names.")
else:
    print("No duplicated elements remain after filtering for _fwd_ names.")

# combine the deduplicated elements with the non-duplicated elements:
ref_only_element_metadata_table_with_measurements_without_duplicates = ref_only_element_metadata_table_with_measurements.loc[~ref_only_element_metadata_table_with_measurements.duplicated(subset=['chr', 'start', 'end'], keep=False)].copy()
ref_only_element_metadata_table_with_measurements_no_duplicates = pd.concat([ref_only_element_metadata_table_with_measurements_deduplicated, ref_only_element_metadata_table_with_measurements_without_duplicates], ignore_index=True)
ref_only_element_metadata_table_with_measurements_no_duplicates
```

```python
# filter for active vs non active elements and write the bed files and print the number of elements in each category:
elements_not_significant = ref_only_element_metadata_table_with_measurements_no_duplicates.loc[~ref_only_element_metadata_table_with_measurements_no_duplicates['is_significant']].copy()
print(f"Number of non-significantly active elements: {elements_not_significant.shape[0]}")
more_active_elements = ref_only_element_metadata_table_with_measurements_no_duplicates.loc[ref_only_element_metadata_table_with_measurements_no_duplicates['is_significant'] & (ref_only_element_metadata_table_with_measurements_no_duplicates['raw_activity_NGN2_log2FC_bcalm'] > 0)].copy()
print(f"Number of significantly more active elements: {more_active_elements.shape[0]}")

less_active_elements = ref_only_element_metadata_table_with_measurements_no_duplicates.loc[ref_only_element_metadata_table_with_measurements_no_duplicates['is_significant'] & (ref_only_element_metadata_table_with_measurements_no_duplicates['raw_activity_NGN2_log2FC_bcalm'] < 0)].copy()
print(f"Number of significantly less active elements: {less_active_elements.shape[0]}")

active_elements = pd.concat([more_active_elements, less_active_elements], ignore_index=True)
print(f"Total number of significantly active elements: {active_elements.shape[0]}")
```

```python
active_elements
```

```python
# are there any duplicates:
if active_elements.loc[active_elements.duplicated(subset=['chr', 'start', 'end'], keep=False)].shape[0] != 0:
    raise ValueError("There are still duplicated elements in the active elements.")
else:
    print("No duplicated elements remain in the active elements.")

if elements_not_significant.loc[elements_not_significant.duplicated(subset=['chr', 'start', 'end'], keep=False)].shape[0] != 0:
    raise ValueError("There are still duplicated elements in the non-significantly active elements.")
else:
    print("No duplicated elements remain in the non-significantly active elements.")
```

```python
import os
output_directory = "/home/kisa/coding/80K_MPRA/80K-Analysis/06_variant_element_analysis/smk_element_driver_enrichment/resources"
output_name_active = "NGN2_scramble_ctrl_active_elements.bed.gz"
output_name_inactive = "NGN2_scramble_ctrl_inactive_elements.bed.gz"
writing=False
if writing:
    active_elements[['chr', 'start', 'end', 'name']].to_csv(os.path.join(output_directory, output_name_active), sep="\t", header=False, index=False, compression="gzip")
    elements_not_significant[['chr', 'start', 'end', 'name']].to_csv(os.path.join(output_directory, output_name_inactive), sep="\t", header=False, index=False, compression="gzip")
```

#### Between considered open and closed active sequences:

```python
active_considered_open_acc_SCREEN = active_elements.loc[active_elements['is_considered_open_acc_SCREEN']].copy()
print(f"Number of active elements that are considered open in SCREEN: {active_considered_open_acc_SCREEN.shape[0]} ({(active_considered_open_acc_SCREEN.shape[0] / active_elements.shape[0]) * 100:.2f}%)")
active_considered_closed_acc_SCREEN = active_elements.loc[~active_elements['is_considered_open_acc_SCREEN']].copy()
print(f"Number of active elements that are considered closed in SCREEN: {active_considered_closed_acc_SCREEN.shape[0]} ({(active_considered_closed_acc_SCREEN.shape[0] / active_elements.shape[0]) * 100:.2f}%)")
```

```python
output_name_active_open = "NGN2_scramble_ctrl_active_elements_considered_open_acc_SCREEN.bed.gz"
output_name_active_closed = "NGN2_scramble_ctrl_active_elements_considered_closed_acc_SCREEN.bed.gz"
writing=False
if writing:
    active_considered_open_acc_SCREEN[['chr', 'start', 'end', 'name']].to_csv(os.path.join(output_directory, output_name_active_open), sep="\t", header=False, index=False, compression="gzip")
    active_considered_closed_acc_SCREEN[['chr', 'start', 'end', 'name']].to_csv(os.path.join(output_directory, output_name_active_closed), sep="\t", header=False, index=False, compression="gzip")
```
