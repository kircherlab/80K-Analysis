```python
def jaccard_index_binary_columns(df, col1, col2):
    """
    Computes the Jaccard index between two binary columns of a DataFrame.

    Args:
        df (pd.DataFrame): The input DataFrame.
        col1 (str): The name of the first binary column.
        col2 (str): The name of the second binary column.

    Returns:
        float: The Jaccard index. Returns 0.0 if the union is zero.
    """
    # Ensure columns contain only 0s and 1s (optional, but good for validation)
    if not df[col1].isin([0, 1]).all() or not df[col2].isin([0, 1]).all():
        raise ValueError("Columns must contain only 0s and 1s.")

    # Calculate the intersection (both columns have 1)
    intersection = ((df[col1] == 1) & (df[col2] == 1)).sum()

    # Calculate the union (at least one column has 1)
    union = ((df[col1] == 1) | (df[col2] == 1)).sum()

    if union == 0:
        return 0.0  # Handle the case where there are no 1s in either column

    jaccard = intersection / union
    return jaccard

jaccard_index_binary_columns(active_genes_table, "Excitatory neurons", "Inhibitory neurons")
```

```python
def from_masterlist_info_2_gene_symbol(masterlist_info):
    """Expects to get a dict as input and tried to read the human gene_symbol"""
    try:
        gene_symbol = masterlist_info['species']['HUMAN']['gene_symbol']
        uniprot_id =  masterlist_info['species']['HUMAN']['uniprot_id']
        uniprot_accession_num =  masterlist_info['species']['HUMAN']['uniprot_ac']
    except:
        print("Problem during finding gene symbol of the following masterlist info: ", masterlist_info)
        gene_symbol = pd.NA
    return pd.Series({'gene_symbol': gene_symbol, 'uniprot_id': uniprot_id, 'uniprot_accession_num': uniprot_accession_num})


# binary_gene_activity_matrix_path = "/home/kisa/coding/TF_Project_cell_type_specific/Results/gene_matrix_specificity_by_percentage.tsv.gz"
# binary_gene_activity_matrix_df = pd.read_csv(binary_gene_activity_matrix_path, sep="\t", low_memory=False)
# # get all genes which are considered active within neurons:
# binary_gene_activity_matrix_df['neuron_active'] = (binary_gene_activity_matrix_df["Excitatory neurons"] == 1) | (binary_gene_activity_matrix_df["Inhibitory neurons"] == 1)

hocomoco_annotation_info_path = "/home/kisa/coding/80K_MPRA/fimo_data/H13CORE-CLUSTERED_annotation.jsonl"
hocomoco_annotation_info_all = pd.read_json(path_or_buf=hocomoco_annotation_info_path, lines=True)
hocomoco_annotation_info_all
# rename name
hocomoco_annotation_info = hocomoco_annotation_info_all.rename(columns={"name": "motif_id"}).copy()

hocomoco_annotation_info[['gene_symbol', 'uniprot_id', 'uniprot_accession_num']] = hocomoco_annotation_info["masterlist_info"].apply(from_masterlist_info_2_gene_symbol)
# only name and tf name
hocomoco_annotation_info = hocomoco_annotation_info[['motif_id', 'gene_symbol', 'uniprot_id', 'uniprot_accession_num', 'cluster_motifs']].copy()

hocomoco_annotation_info['is_TFBS'] = 1
hocomoco_annotation_info
```

```python
def get_if_cell_type_specific_or_agnostic(row, new_column, cell_type_count_col, upper_threshold=0.75, lower_threshold=0.1):
    """
    Determines if a gene is cell type specific or agnostic based on two thresholds upper and lower.

    """
    activity_type = "intermediate"
    if row[cell_type_count_col] >= upper_threshold:
        activity_type = "agnostic"
    elif row[cell_type_count_col] <= lower_threshold:
        activity_type = "specific"
    row[new_column] = activity_type
    return row

upper_percentile = expressed_genes_in_cell_types_TFBS['cell_type_number'].quantile(0.75) # 81.0
lower_percentile = expressed_genes_in_cell_types_TFBS['cell_type_number'].quantile(0.05) # 15
lower_percentile = expressed_genes_in_cell_types_TFBS['cell_type_number'].quantile(0.1) # 30.0

expressed_genes_in_cell_types_TFBS = expressed_genes_in_cell_types_TFBS.apply(lambda row: get_if_cell_type_specific_or_agnostic(row, new_column='activity_type', cell_type_count_col='cell_type_number', upper_threshold=upper_percentile, lower_threshold=lower_percentile), axis=1)
upper_threshold = 80
lower_threshold = 30
expressed_genes_in_cell_types_TFBS = expressed_genes_in_cell_types_TFBS.apply(lambda row: get_if_cell_type_specific_or_agnostic(row, new_column='activity_type', cell_type_count_col='cell_type_proportion', upper_threshold=upper_threshold, lower_threshold=lower_threshold), axis=1)
expressed_genes_in_cell_types_TFBS
```
