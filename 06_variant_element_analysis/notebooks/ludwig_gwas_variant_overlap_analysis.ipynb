```python
import pandas as pd
import matplotlib.pyplot as plt

VARIANT_ID_COL = 'SPDI_elem'
PHENOTYPE_CODE_COL_VARIANTS = 'source_file' # Column in variants_data.tsv with phenotype codes
PHENOTYPE_CODE_COL_PHENOTYPES = 'Phenotype Code' # Column in phenotype_table.tsv with phenotype codes
PHENOTYPE_DESC_COL = 'Phenotype Description'

# filter for only significant MPRA variants
sig_gwas_variants_df = gwas_variant_df_raw_merged[gwas_variant_df_raw_merged['is_significant_adjusted_pval']].copy()


# 2. Count phenotype associations per variant
# Each row in sig_gwas_variants_df represents one variant-phenotype association.
# We need to count distinct phenotype associations for each variant.
# First, ensure we're counting unique variant-phenotype pairs, not just rows for a variant.
# We'll consider a unique association as a unique (SPDI_elem, phenotype_description) pair.
unique_associations_df = sig_gwas_variants_df[[VARIANT_ID_COL, PHENOTYPE_DESC_COL]].drop_duplicates()
variant_associations_count = unique_associations_df[VARIANT_ID_COL].value_counts()

# 3. Get the top 10 variants
top_10_variants = variant_associations_count.head(10)

# 4. Calculate overall significant variant associations
# This usually refers to variants that have *at least one* phenotype association.
# If your variants_data.tsv only contains "significant" associations, then the number
# of unique variants in that file is the "overall significant variant associations."
overall_significant_variant_associations = gwas_variant_df_raw_merged[VARIANT_ID_COL].nunique()

# --- Plotting ---
plt.figure(figsize=(12, 7))
top_10_variants.plot(kind='bar', color='skyblue')

plt.title(
    f'Number of Phenotype Associations for Top 10 Variants\n'
    f'(Overall Significant Variant Associations: {overall_significant_variant_associations})',
    fontsize=14,
    pad=20
)
plt.xlabel('Variant (SPDI_elem)', fontsize=12)
plt.ylabel('Number of Unique Phenotype Associations', fontsize=12)
plt.xticks(rotation=45, ha='right') # Rotate labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

print("\nTop 10 Variants by Number of Phenotype Associations:")
print(top_10_variants)
print(f"\nOverall number of significant variant associations: {overall_significant_variant_associations}")
```
