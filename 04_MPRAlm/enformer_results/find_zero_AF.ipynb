## Check vcf files for variant with zero AF

```python
import pandas as pd
import pysam
import os
import yaml

# config 
config_path = "/data/gpfs-1/users/kisa11_c/work/coding/MPRA/kilian_projects/80K_MPRA_variants/config.yaml"
with open(config_path) as conf:
    config = yaml.load(conf, Loader=yaml.FullLoader)
    conf.close()
```

```python
# fix header (columns need to be in other order)
# load tsv with pysam go through info column
name = config["names"][3] + config["file_ending"]
file_name = os.path.join(config["ultra_rare_variants_directory"], name)
# is not valid vcf: read pandas file (\t)
def mohan_to_vcf_file(file_name):
    """Get mohan format: ['#CHROM', 'POS', 'ID', 'REF', 'ALT', 'DNase_max', 'INDEX_max', 'QUAL', 'FILTER', 'INFO'] into vcf format"""
vcf_like = pd.read_csv(file_name, sep="\t")
vcf_like["FORMAT"] = [" "] * vcf_like.shape[0]
column_list = vcf_like.columns.to_list()
new_order = [0, 1, 2, 3, 4, 7, 8, 9, 10]
new_column_list = [column_list[i] for i in new_order]

vcf_like = vcf_like[new_column_list]

# write to csv
outfile = "/data/gpfs-1/users/kisa11_c/work/coding/MPRA/kilian_projects/80K_MPRA_variants/enformer_results/new_cardiac.ultra-rare_variants.vcf"
vcf_like.to_csv(outfile, sep="\t")
vcf_like
```

```python
overall_count = 0
for name in config["names"]:
    file_name = name + config["file_ending"]

    file_path = os.path.join(config["ultra_rare_variants_directory"], name, "variants", file_name)
    vcf_in = pysam.VariantFile(file_path, 'r')


    zero_af_count = 0
    for rec in vcf_in.fetch():
        overall_count += 1
        if (rec.info["AF"][0] == 0):
            zero_af_count += 1
            # print("found Variant with AF=0")


    print(f"done with {name}, found {zero_af_count} variants with AF==0")
print(f"Done with looking at the files and checked {overall_count} variants")
```
