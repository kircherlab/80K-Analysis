````python
def extract_reference_from_kircher_name(name):
    """Gets kircher header
    1. if GC_Kircher:ALT_:
        remove the last underscore

    remove the first two underscores
    remove all occurences of KircherControls
    and return as reference name
    """
    if name.startswith("GC_Kircher:ALT_"):
        name = "_".join(name.split("_")[:-5])
    region_name = "_".join(name.split("_")[2:])
    return region_name


def extract_variant_info_from_kircher_name(name):
    if name.startswith("GC_Kircher:ALT_"):
        return "_".join(name.split("_")[-5:])
    return pd.NA


def shorten_kircher_control_name(row, tile_col="tile_info", variant_col="variant_info", name_col="name", chr_col="chr", start_col="start", end_col="end"):
    """
    Remove the region info and only have GC_Kircher:REF_<tile_info> or GC_Kircher:ALT_<tile_info>_variant_info
    """
    name = row[name_col]
    tile_info = row[tile_col]
    ref_alt_info = "_".join(name.split("_")[:2])
    region_info = f'{row[chr_col]}:{row[start_col]}-{row[end_col]}'
    new_name = f'{ref_alt_info}_SORT1_{region_info}_{tile_info}' # only SORT1 regions santiy checked with Max
    if "GC_Kircher:ALT" in ref_alt_info:
        variant_info = row[variant_col]
        return f"{new_name}_{variant_info}"
    return new_name

````

````python
# gather info
general_kircher['reference'] = general_kircher[col_name].apply(extract_reference_from_kircher_name)
general_kircher['variant_info'] = general_kircher[col_name].apply(extract_variant_info_from_kircher_name)
general_kircher["tile_info"] = general_kircher["reference"].apply(lambda ref: ref.split("_")[-1])

# shorten the name
general_kircher["short_name"] = general_kircher.apply(lambda row: shorten_kircher_control_name(row, tile_col="tile_info", variant_col="variant_info", name_col="name", chr_col="chr", start_col="start", end_col="end"), axis=1)
````

````python
def get_control_class_based_on_name_or_label(row, class_col="class", name_col="name", label_col="label", category_col="category"):
    """The given sequence is classified as "element inactive control" if the name contains scramble or the category is scrambled or the label contains _negative_"""
    if row[class_col] != "element inactive control":
        return row
    row[class_col] = "element active control"
    if row[category_col] == "scrambled" or ("negative" in row[label_col]) or ("scramb" in row[name_col]):
        row[class_col] = "element inactive control"
    return row
````

````python
import re
def get_mk_region_from_header(row, name_col="name", chr_col="chr", start_col="start", strand_col="strand", end_col="end"):
    """

    """
    header = row[name_col]
    # Regex to capture the chromosome, start, and end
    pattern = re.compile(r"(chr\w+)-(\d+)([+-])(\d+)")

    match = pattern.search(header)
    if match:
        chrom, start, strand, end = match.groups()
        row[chr_col] = chrom
        row[start_col] = int(start)-1
        row[strand_col] = strand
        row[end_col] = int(end)
    else:
        print(header)
    return row


def add_SNV_or_indel_for_variant_header(row, name_col="name", class_col="class", category_col="category", variant_class_col="variant_class", variant_pos_col="variant_pos", info_col="info"):
    header = row[name_col]
    if not header.startswith("MK"):
        raise ValueError("Function only defined for the MK group")
    pattern_for_SNV = re.compile(r"(\w)-(\w)-(\d+)")
    pattern_for_indel_with_numbers = re.compile(r"ts(\d+)-(\d+)")
    pattern_for_lc_indel = re.compile(r"LC(\d+).(\d)+")
    SNV_match = pattern_for_SNV.search(header)
    indel_with_numbers_match = pattern_for_indel_with_numbers.search(header)
    indel_LC_match = pattern_for_lc_indel.search(header)
    row[category_col] = "element"
    row[class_col] = "element active control"
    if "reference" in header:
        row[category_col] = "element"
        row[class_col] = "element active control"

    elif "scramble_negative" in header:
        row[category_col] = "scrambled"
        row[class_col] = "element inactive control"

    elif SNV_match:
        row[category_col] = "variant"
        row[class_col] = "variant positive control"
        REF, ALT, variant_pos_one_based = SNV_match.groups()
        row[variant_class_col] = "SNV"
        row[variant_pos_col] = int(variant_pos_one_based) - 1 # from 1-based to 0-based

    elif indel_with_numbers_match:
        row[category_col] = "variant"
        row[class_col] = "variant positive control"
        start, end = indel_with_numbers_match.groups()
        row[variant_class_col] = "indel"
        row[variant_pos_col] = int(start) - 1
        row[info_col] = "for start and end (1-based) of the variant see header"

    elif indel_LC_match:
        row[category_col] = "variant"
        row[class_col] = "variant positive control"
        row[variant_class_col] = "indel"
        row[variant_pos_col] = pd.NA
        row[info_col] = "The meaning of LC was not known while generating this file"



    else:
        print(header)
        raise ValueError("Unexprected Sequence this function expected LC containing ts containing or SNVs")
    return row

def correct_MK_metadata(row):
    """
    Correct MK metadata:
    - category treat everything as element except the ones which contain scramble in the header => scrambled
    - set all as element active control except the ones which contain scramble => element inactive control
    - get the regions from the header (start -1) and strand from inbetween
    """
    pass
````
