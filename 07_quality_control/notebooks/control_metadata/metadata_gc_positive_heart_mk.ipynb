```python
def get_allele_from_sequence(sequence, ref, alt, variant_position):
    """Checks if a sequences has at the variant position ref or alt"""
    if sequence[variant_position] == ref:
        return 'ref'
    elif sequence[variant_position] == alt:
        return 'alt'
    else:
        raise ValueError("Sequence is not ref and not alt according to the given information")

# dict of chr number to refseq chromosome number
chrom_2_refseq = {"chr1": "NC_000001.11",
    "chr2": "NC_000002.12",
    "chr3": "NC_000003.12",
    "chr4": "NC_000004.12",
    "chr5": "NC_000005.10",
    "chr6": "NC_000006.12",
    "chr7": "NC_000007.14",
    "chr8": "NC_000008.11",
    "chr9": "NC_000009.12",
    "chr10": "NC_000010.11",
    "chr11": "NC_000011.10",
    "chr12": "NC_000012.12",
    "chr13": "NC_000013.11",
    "chr14": "NC_000014.9",
    "chr15": "NC_000015.10",
    "chr16": "NC_000016.10",
    "chr17": "NC_000017.11",
    "chr18": "NC_000018.10",
    "chr19": "NC_000019.10",
    "chr20": "NC_000020.11",
    "chr21": "NC_000021.9",
    "chr22": "NC_000022.11",
    "chrX": "NC_000023.11",
    "chrY": "NC_000024.10"}

def get_SPDI(chrom, start, variant_pos_zero_based, sequence, ref, alt):
    """Check if variant position is the given alternative and generate SPDI"""
    if get_allele_from_sequence(sequence=sequence, ref=ref, alt=alt, variant_position=variant_pos_zero_based) != 'alt':
        raise ValueError("Expected to generate SPDI only for alternative sequences")
    return f"{chrom_2_refseq[chrom]}:{start + variant_pos_zero_based}:{ref}:{alt}"


def get_coordinates_control_pos_neg_heart_MK(row):
    """
    Special cases to set chr, start, end and strand for control sequences from their header (because not in region bed)
    Case: C_negative_heart_MK: ( all C_negative_heart_MK sequences have "_chr" pattern)
        header: C_negative_heart_MK:tile_6903_chr11_9614045_9614314_reference__0.958461950470297
            => chr: 11, start: 9614045, end: 9614314, strand: . (no information given)

    Case: C_positive_heart_MK: (all C_positive_heart_MK sequences have "_chr" pattern)
        header: C_positive_heart_MK:tile_7939_chr11_65487592_65487861_reference_1.25449216981846
            => chr: 11, start: 65487592, end: 65487861, strand: . (no information given)

    """
    name = row[col_name]
    if name.startswith('C_negative_heart_MK') or name.startswith('C_positive_heart_MK'): # or name.startswith('C_positive_neuron_MK') or name.startswith('C_negative_neuron_MK'):
        if 'scramble' in name:
            row[col_category] = 'scrambled'
            return row

        row[col_class] = 'element inactive control'
        if 'positive_neuron' in name:
            row[col_class] = 'element active control'
        region_info = '_'.join(name.split('_chr')[1].split('_')[:3]) # 11_9614045_9614314
        row[col_category] = 'element'
        row[col_chr] = f"chr{region_info.split('_')[0]}"
        row[col_start] = int(region_info.split('_')[1]) - 1 # turn into 0-based
        row[col_end] = int(region_info.split('_')[2])
        row[col_strand] = '.'
        if len(name.split('_chr')[1].split("_")) == 7:
            row[col_variant_class] = 'SNV'
            row[col_category] = 'variant'
            row[col_class] = 'variant negative control'
            row[my_col_ref_base] = name.split('_chr')[1].split("_")[3]
            row[my_col_alt_base] = name.split('_chr')[1].split("_")[4]
            row[col_variant_pos] = int(name.split('_chr')[1].split("_")[5]) - 1
            row[col_SPDI] = get_SPDI(chrom=row[col_chr], start=row[col_start], variant_pos_zero_based=row[col_variant_pos], sequence=row[col_sequence], ref=row[my_col_ref_base], alt=row[my_col_alt_base])
            row[col_allele] = get_allele_from_sequence(row[col_sequence], row[my_col_ref_base], row[my_col_alt_base], row[col_variant_pos])
            if 'positive_neuron' in name:
                row[col_class] = 'variant positive control'
    return row
```

```python
# function to generate arrays out off the columns
def make_column_arrays(row):
    """
    create arrays for the required columns
    """
    allele = row[col_allele]
    SPDI = row[col_SPDI]
    variant_pos = row[col_variant_pos]
    variant_class = row[col_variant_class]

    if isinstance(allele, str): # only "alt" is string
        row[col_allele] = [allele]
    if isinstance(SPDI, str): # only for alt sequences this is true
        row[col_SPDI] = [SPDI]
    if isinstance(variant_pos, float):
        row[col_variant_pos] = [int(variant_pos)]
    elif isinstance(variant_pos, int):
        row[col_variant_pos] = [int(variant_pos)]
    if isinstance(variant_class, str):
        if row[col_variant_class] in ['SNV', 'indel']:
                row[col_variant_class] = [variant_class]
    if not isinstance(row[col_class], str):
        print(row[col_name])
    return row
```
