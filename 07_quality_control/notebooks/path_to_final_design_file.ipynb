#### Help from Pia to clean the duplicates with focus on the variants (not mixing ref and alt up because of the variant map this would lead to losing the variant)
- I thought there is a problem in the final design with the variants and I can do this with clumpsify but the more robust way is to use the variant map:
  - find for the header the fitting alternative ID
    - Split the controls and get all variants 
    - Split controls into ref and alt
    - for each alt there needs to be an entry in the variant map 
    - for each reference associated to the alternative look into the set of references if it is in there
    

````python
def ignore_this_control_header(header):
    """
    Checks if the given header should be ignored
    """
    label = hf.get_label(header)
    # if MK => not interesting
    if label == "MK":
        return True
    return False
def is_control_variant_reference(header):
    """Returns if a control header is a variant of a reference"""
    if (not ignore_this_control_header):
        return False

    # currently best pattern for variants (REF_ might be not strict enough)
    if ("REF_" in header) or is_control_alternative(header):
        return True
    return False


def is_control_reference(header):
    """Returns if a control header is a control reference sequence. Some control groups might be ignored"""
    if (not ignore_this_control_header):
        return False
    # check reference pattern
    if "REF_" in header:
        return True
    return False


def is_control_alternative(header):
    """Returns if a control header is a control alternative sequence. Some control groups might be ignored"""
    if (not ignore_this_control_header):
        return False
    # check alternative pattern
    if "ALT_" in header:
        return True
    return False
````

````python
# find for the header the fitting alternative ID
# Split the controls and get all variants
# final_design_controls_df['header'].to_csv('final_controls_find_variant_pattern.csv', sep="\t", index=False)

# make headers matchable
final_design_controls_df['matchable_header'] = final_design_controls_df['header'].str.replace('~',',')
final_design_controls_df['matchable_header'] = final_design_controls_df['matchable_header'].str.replace('*','>')

only_control_variants = final_design_controls_df.loc[final_design_controls_df['header'].apply(is_control_variant_reference)]
# only_control_variants # 967

# Split controls into ref and alt
only_control_alt = only_control_variants.loc[only_control_variants['header'].apply(is_control_alternative)]
only_control_alt # 655 => 967 - 655 = 312 (references)

# all reference headers in the control group
only_control_ref = only_control_variants.loc[only_control_variants['header'].apply(is_control_reference)]
only_control_ref # 312

````
