### Investigate which and how many oligos are matched

```python
def fasta_to_tsv(fasta_file, tsv_file):
    with open(fasta_file, "r") as f:
        with open(tsv_file, "w") as t:
            for line in f:
                if line.startswith(">"):
                    t.write(line[1:].strip())
                    t.write("\t")
                else:
                    t.write(line.strip())
                    t.write("\n")
```

```python
def plot_oligo_label_distribution(oligo_label_df, outpath):
    """
    Plots the distribution of the labels in the dataframe
    @input: label_df: dataframe with a column "label" and a column "oligo"
    """
    sorted_missing_seqs = oligo_label_df.groupby('label').count().sort_values(by='oligo', ascending=False)["oligo"]
    ax = sorted_missing_seqs.plot(kind='bar', figsize=(20, 6))
    print(ax.patches)
    for p in ax.patches:
        ax.annotate(str(p.get_height()), (p.get_x() * 1.005, (p.get_height() + 15) * 1.005))
    # store the plot as png
    ax.get_figure().savefig(outpath, dpi=300, bbox_inches='tight')
    plt.close()
 
```

```python
output_dir = "/data/gpfs-1/users/kisa11_c/work/coding/80K_analysis/01_missing_sequences"
# plot the distribution of the found oligos
plot_oligo_label_distribution(found_oligos, os.path.join(output_dir , "identified_oligos/found_oligos_distribution.png"))

# # plot the distribution of the found oligos in the reverse reads
plot_oligo_label_distribution(reverse_found_oligos, os.path.join(output_dir , "identified_oligos/reverse_found_oligos_distribution.png"))
# # plot the distribution of the missing oligos
plot_oligo_label_distribution(still_missing_oligos, os.path.join(output_dir , "identified_oligos/missing_oligos_distribution.png"))
# # plot the distribution of the found oligos in the forward reads
plot_oligo_label_distribution(forward_found_oligos, os.path.join(output_dir , "identified_oligos/forward_found_oligos_distribution.png"))
```
